## Association Rule Mining (ARM)

#### Definition
Association Rule Mining is a technique used to discover relationships between items in a dataset.

It is mainly used in Market Basket Analysis.

---

#### Example
If a customer buys Bread → they may also buy Butter

---

#### Key Terms

1. Support
- It shows how frequently an item appears in dataset
- Formula:
  Support(A) = (Transactions containing A) / (Total Transactions)

2. Confidence
- It shows how often the rule is true
- Formula:
  Confidence(A → B) = Support(A ∪ B) / Support(A)

3. Lift
- It shows strength of rule
- Formula:
  Lift(A → B) = Confidence(A → B) / Support(B)

- Lift > 1 → Strong rule
- Lift = 1 → No relation
- Lift < 1 → Weak rule

---

#### Steps in Association Rule Mining

1. Collect transaction dataset
2. Find frequent itemsets using Apriori
3. Apply minimum support threshold
4. Generate rules using confidence
5. Evaluate rules using lift

---

#### Algorithm Used
- Apriori Algorithm (most common)

---

#### Advantages
- Helps in product recommendation
- Easy to understand
- Useful in business decision making

---

#### Disadvantages
- Computationally expensive
- Generates too many rules
- Needs proper threshold selection

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
# Load dataset
df = pd.read_csv("groceries.csv")

df.head()
```

```python
# Convert each row into list (remove NaN values)

transactions = []

for i in range(len(df)):
    transactions.append([str(item) for item in df.iloc[i] if pd.notna(item)])

# Check first 5 transactions
transactions[:5]
```

```python
# Apply Transaction Encoder
# used because to converts transactions into binary format which is required for apriori algorithm

te = TransactionEncoder()
te_data = te.fit(transactions).transform(transactions)

basket = pd.DataFrame(te_data, columns=te.columns_)

basket.head()
```

## Apriori Algorithm

#### Definition
Apriori Algorithm is used in Association Rule Mining to find frequent itemsets and generate association rules using support, confidence, and lift.

---

### Basic Concept (Apriori Property)
If an itemset is frequent, then all of its subsets must also be frequent.

Example:
If {Milk, Bread} is frequent, then {Milk} and {Bread} must also be frequent.

---

#### Key Terms

1. Itemset
A collection of items (e.g., Milk, Bread)

2. Frequent Itemset
Itemset whose support is greater than minimum threshold

3. Support
Support(A) = (Transactions containing A) / (Total Transactions)

---

### Steps of Apriori Algorithm

Step 1: Find 1-itemsets
- Calculate support of each item
- Keep only those above minimum support

Step 2: Generate 2-itemsets
- Combine frequent 1-itemsets
- Calculate their support

Step 3: Pruning
- Remove itemsets with low support
- If subset is not frequent, discard full itemset

Step 4: Generate higher itemsets (3-itemsets, etc.)
- Repeat combine + support + prune

Step 5: Stop Condition
- Stop when no new itemsets are generated

Step 6: Generate Rules
- Create rules from frequent itemsets
- Calculate confidence and lift

---

#### Flow
Generate → Test → Prune → Repeat

---

#### Advantages
- Easy to understand
- Reduces search space using pruning
- Useful in market basket analysis

---

#### Disadvantages
- Slow for large datasets
- Requires multiple scans
- High memory usage

```python
# Find frequent itemsets

frequent_items = apriori(basket, min_support=0.02, use_colnames=True)

frequent_items.head()
```

```python
# Generate association rules

rules = association_rules(frequent_items, metric="lift", min_threshold=1)

rules.head()
```
