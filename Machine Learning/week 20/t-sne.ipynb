## t-SNE (t-Distributed Stochastic Neighbor Embedding)

#### Definition
t-SNE is a dimensionality reduction technique used to visualize high-dimensional data in 2D or 3D.

---

#### Why t-SNE?

Real-world data has many features (high dimensions).
t-SNE helps convert it into 2D so we can visualize it easily.

---

#### Simple Idea

- Similar data points stay close
- Different data points stay far

---

#### Working

1. Calculate similarity between points in high dimension
2. Map them into low dimension (2D/3D)
3. Try to preserve distances

---

#### Advantages
- Good for visualization
- Shows clusters clearly

---

#### Disadvantages
- Slow for large data
- Not useful for prediction
- Results may change each time

---

#### Use Case
- Data visualization
- Cluster understanding

```python
# Import required libraries

import numpy as np                 # For numerical operations
import matplotlib.pyplot as plt    # For plotting graph
from sklearn.datasets import load_iris   # Sample dataset
from sklearn.manifold import TSNE        # t-SNE algorithm
```

```python
# Load Iris dataset (flower dataset)

data = load_iris()

# Features (independent variables)
X = data.data

# Labels (target - type of flower)
y = data.target

# Check shape of data
print("Shape of X:", X.shape)
```

```python
# Create t-SNE model

tsne = TSNE(
    n_components=2,    # Convert data into 2 dimensions (2D)
    random_state=0     # For reproducible results (same output every time)
)

# Fit and transform the data
X_tsne = tsne.fit_transform(X)

# Check transformed data shape
print("Shape after t-SNE:", X_tsne.shape)
```

```python
# Plot the t-SNE result

plt.figure(figsize=(8,6))

# Scatter plot
plt.scatter(
    X_tsne[:, 0],    # X-axis (component 1)
    X_tsne[:, 1],    # Y-axis (component 2)
    c=y              # Color based on class (flower type)
)

# Add labels and title
plt.title("t-SNE Visualization of Iris Dataset")
plt.xlabel("Component 1")
plt.ylabel("Component 2")

# Show plot
plt.show()
```
