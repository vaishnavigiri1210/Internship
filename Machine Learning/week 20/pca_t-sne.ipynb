#### Difference Between PCA and t-SNE

| Feature | PCA (Principal Component Analysis) | t-SNE (t-Distributed Stochastic Neighbor Embedding) |
|--------|------------------------------------|----------------------------------------------------|
| Type | Linear technique | Non-linear technique |
| Goal | Maximize variance | Preserve similarity between points |
| Purpose | Data reduction + preprocessing | Data visualization |
| Speed | Fast | Slow |
| Output | Stable (same result every time) | Changes (randomness involved) |
| Structure Preserved | Global structure | Local structure |
| Clusters | May overlap | Clearly separated |
| Use Case | Feature reduction, ML models | Visualization, pattern detection |
| Complexity | Low | High |


#### Conclusion

PCA is used for reducing dimensions while maintaining overall structure,  
whereas t-SNE is used for visualizing data by showing clear clusters.

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
```

```python
# Load Iris dataset
data = load_iris()

X = data.data   # features
y = data.target # labels
```

```python
# Apply PCA (convert to 2D)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
```

```python
# Apply t-SNE (convert to 2D)

tsne = TSNE(n_components=2, random_state=0)
X_tsne = tsne.fit_transform(X)
```

```python
plt.figure(figsize=(6,5))

plt.scatter(X_pca[:,0], X_pca[:,1], c=y)

plt.title("PCA Visualization")
plt.xlabel("PC1")
plt.ylabel("PC2")

plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.scatter(X_tsne[:,0], X_tsne[:,1], c=y)

plt.title("t-SNE Visualization")
plt.xlabel("Component 1")
plt.ylabel("Component 2")

plt.show()
```
