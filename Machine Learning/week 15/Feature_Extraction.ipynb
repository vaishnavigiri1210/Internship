### Feature Extraction

Feature extraction means transforming raw data into a set of meaningful features that improves model performance.

👉 Instead of selecting existing features (like in feature selection), here we generate new features.

Example

| Height (cm) | Weight (kg) |
| ----------- | ----------- |
| 170         | 65          |

👉 Feature Extraction can create a new feature:

BMI = weight / (height²)

Now model gets:

Height, Weight, BMI (new feature)

### Why Feature Extraction is Important?  

Reduces Dimensionality  
Converts large data into fewer features  
Removes Noise  
Keeps only useful information  
Improves Accuracy  
Better representation → better predictions  
Handles Complex Data  
Useful for images, text, audio  

### Types of Feature Extraction 

1. Statistical Methods 
 
Mean, Median, Variance  
Used in numeric data  

👉 Example : From signal data → extract average value

2. Dimensionality Reduction Techniques  

PCA (Principal Component Analysis)  
Converts features into principal components  
Keeps maximum variance  

👉 Used when data has many features

LDA (Linear Discriminant Analysis)  
Focuses on class separation  

👉 Used in classification problems  

3. Text Feature Extraction

Used in NLP:

Bag of Words (BoW)  
TF-IDF (Term Frequency – Inverse Document Frequency)  

👉 Converts text → numbers

4. Image Feature Extraction
Edge detection  
Color features  
CNN (Deep Learning)  

👉 Used in image recognition  
5. 🔸 Signal/Audio Feature Extraction  
MFCC (Mel Frequency Cepstral Coefficients)  

👉 Used in speech recognition

```python
import numpy as np                  # for numerical operations
import pandas as pd                 # for handling dataset
import matplotlib.pyplot as plt     # for visualization
```

```python
from sklearn.datasets import load_iris   # sample dataset
```

```python
# Step 2: Load dataset
iris = load_iris()          # load iris dataset
X = iris.data              # features (4 columns)
y = iris.target            # target labels
```

```python
df = pd.DataFrame(X, columns=iris.feature_names)
df['target'] = y
```

```python
print("Original Data Shape:", X.shape)  # (150, 4)
```

```python
from sklearn.decomposition import PCA    # PCA for feature extraction
from sklearn.preprocessing import StandardScaler  # for scaling
```

```python
# Step 4: Feature Scaling (VERY IMPORTANT before PCA)
scaler = StandardScaler()     # create scaler object
X_scaled = scaler.fit_transform(X)  # scale data
```

```python
# Step 5: Apply PCA (Feature Extraction)
pca = PCA(n_components=2)     # reduce to 2 features
X_pca = pca.fit_transform(X_scaled)

print("New Data Shape after PCA:", X_pca.shape)  # (150, 2)
```

```python
# Step 6: Explained Variance (how much information retained)
print("Explained Variance Ratio:", pca.explained_variance_ratio_)
```

```python
# Step 7: Visualization
plt.figure()

# Scatter plot of new features
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Feature Extraction")

plt.show()
```
