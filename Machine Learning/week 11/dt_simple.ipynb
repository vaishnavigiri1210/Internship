```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('movies.csv')
df
```

```python
df.shape
```

```python
df.columns
```

```python
sns.jointplot(x = 'Budget', y = 'Celebrities', data = df,
             hue = 'Class');
```

```python
x = df[['Budget', 'Celebrities']]       # input features (independent variables)
y = df['Class']                  # target variable (dependent variable)
```

```python
from sklearn.tree import DecisionTreeClassifier    
```

```python
dt = DecisionTreeClassifier(random_state= 0)        # Create an instance of the Decision Tree Classifier 
# random_state = 0 means same random numbers will be generated every time you run the code, ensuring reproducibility.
```

```python
dt.fit(x, y)        # Fit the model to the data (train the model)
```

```python
dt.predict([[540000000, 3]])    # Predict the class for a movie with a budget of 540 million and 3 celebrities
```

```python
dt.predict([[344000000, 10]])   # Predict the class for a movie with a budget of 344 million and 10 celebrities
```
