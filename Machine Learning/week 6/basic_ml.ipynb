### Machine Learning

Machine Learning (ML) is a part of Artificial Intelligence (AI) that allows a computer to learn from data and improve its performance without being explicitly programmed.

In simple words:
👉 Instead of telling the computer every rule, we give it data, and it finds patterns by itself and makes predictions or decisions.

Rule = condition or logic to make a decision

Like:

if age > 18 → allow vote  
if temperature > 38 → fever  
if balance < 0 → block account  

These are all “rules”.

Suppose you want to predict a student’s marks based on study hours.

Traditional programming:
You write rules like:
if hours > 5 then marks = 80

Machine Learning:
You give the computer many examples of
(study hours → marks)
and it learns the relationship and predicts marks for new students.

### Why ml required

🔹 1. Too many rules to write manually

In real life, problems are very complex.

Example: Spam detection

You cannot write thousands of rules like:

if word = “free” → spam   
if word = “offer” → spam  
if word = “money” → spam  
if word = “lottery” → spam  

New spam messages appear every day. Writing rules for all cases is impossible.

👉 ML learns patterns automatically from data.

🔹 2. Data is very large (Big Data)

Companies have millions of records:

YouTube videos  
Netflix movies  
Bank transactions  
Medical records

Humans cannot analyze such huge data manually.

👉 ML can process and learn from huge datasets quickly.

🔹 3. Problems where rules are not clear

Some problems do not have clear logic.

Example:

Face recognition  
Voice recognition  
Handwriting recognition  

You cannot say:

if nose length = X and eye shape = Y → this is person A

It’s too complex.

👉 ML finds patterns in images and sound automatically.

🔹 4. Systems that improve with experience

ML systems get better with more data.

Example:

Google search results improve over time  
Netflix recommendations become better  
Maps traffic prediction improves daily  
Normal programs stay the same unless a programmer changes them.  

🔹 5. Automation & faster decisions

ML helps in:

Fraud detection in banks  
Disease prediction in hospitals  
Stock market prediction  
Chatbots and assistants  

All these need fast and smart decisions.

🔹 6. Real-world examples

Gmail spam filter  
Face unlock in phones  
Self-driving cars  
Recommendation systems  
Voice assistants

These are not possible with only simple rules.

### Types of ML

✅ 1. Supervised Learning

Data is given with labels (answers)  -> meaning every input has a corresponding correct output. (input,output)  
Model learns from input → output mapping  
Used for prediction & classification

📌 Examples:

Email spam detection  
Marks prediction  
Disease prediction

✅ 2. Unsupervised Learning

Data has no labels (no answers) (only output)  
Model finds patterns & groups by itself  

📌 Examples:

Customer segmentation  
Grouping similar products  
Fraud detection

✅ 3. Semi-Supervised Learning

Data is partly labeled and partly unlabeled  
Combination of supervised + unsupervised

📌 Example:

Image recognition with few labeled images

✅ 4. Reinforcement Learning

Model learns by trial and error  
Gets reward or punishment

📌 Examples:

Game playing (Chess, Ludo, PUBG bots 😄)  
Self-driving cars  
Robot training
