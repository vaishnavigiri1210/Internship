```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('ingredients.csv')
df.head()
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
sns.jointplot(x = 'sweetness', y ='crunchiness', data = df,
             hue = 'class');
```

```python
x = df[['sweetness', 'crunchiness']]        # input features
y = df['class']     # target variable
```

```python
y       # target variable
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors= 4)  
# n_neighbors=4 means that the algorithm will consider the 4 nearest neighbors to make a prediction.
```

```python
knn.fit(x, y)       
# Fit the KNN model to the training data (x and y). 
# This step involves storing the training data so that it can be used for making predictions later.
```

```python
# new data point for prediction

new = pd.DataFrame({
    'sweetness':[6],
    'crunchiness': [4]
})
```

```python
knn.predict(new)
```
