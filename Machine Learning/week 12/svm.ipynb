```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('letterdata.csv')
print(df.head())
```

```python
X = df.drop('letter', axis=1)
y = df['letter']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)
```

| Kernel     | Boundary Type       | Use                     |
| ---------- | ------------------- | ----------------------- |
| Linear     | Straight line       | Simple linear data      |
| Polynomial | Curved              | Moderate complexity     |
| Sigmoid    | Neural network type | Rarely used             |
| RBF        | Complex nonlinear   | Most practical datasets |

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python
model_linear = SVC(kernel='linear')
```

```python
model_linear.fit(X_train, y_train)
```

```python
y_pred_linear = model_linear.predict(X_test)
```

```python
acc_linear = accuracy_score(y_test, y_pred_linear)
```

```python
print("Linear Kernel Accuracy:", acc_linear)
```

```python
model_poly = SVC(kernel='poly', degree=3)
```

```python
model_poly.fit(X_train, y_train)
```

```python
y_pred_poly = model_poly.predict(X_test)
```

```python
acc_poly = accuracy_score(y_test, y_pred_poly)
```

```python
print("Polynomial Kernel Accuracy:", acc_poly)
```

```python
model_sigmoid = SVC(kernel='sigmoid')
```

```python
model_sigmoid.fit(X_train, y_train)
```

```python
y_pred_sigmoid = model_sigmoid.predict(X_test)
```

```python
acc_sigmoid = accuracy_score(y_test, y_pred_sigmoid)
```

```python
print("Sigmoid Kernel Accuracy:", acc_sigmoid)
```

```python
model_rbf = SVC(kernel='rbf')
```

```python
model_rbf.fit(X_train, y_train)
```

```python
y_pred_rbf = model_rbf.predict(X_test)
```

```python
acc_rbf = accuracy_score(y_test, y_pred_rbf)
```

```python
print("RBF Kernel Accuracy:", acc_rbf)
```

```python
kernels = ['Linear', 'Polynomial', 'Sigmoid', 'RBF']
accuracy = [acc_linear, acc_poly, acc_sigmoid, acc_rbf]
```

```python
plt.figure(figsize=(8,5))

sns.barplot(x=kernels, y=accuracy)

plt.title("SVM Kernel Accuracy Comparison")
plt.xlabel("Kernel Type")
plt.ylabel("Accuracy")

plt.show()
```
