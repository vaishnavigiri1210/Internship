## K-Means Clustering

### Introduction
K-Means is an Unsupervised Machine Learning Algorithm used for clustering data into groups.

Unsupervised means:
- Data has no labels
- Algorithm finds patterns on its own

### Definition
K-Means divides the dataset into K number of clusters based on similarity.

#### Working Steps
1. Choose value of K (number of clusters)
2. Initialize K centroids randomly
3. Assign each data point to nearest centroid
4. Update centroid by calculating mean of points
5. Repeat until centroids do not change

### Distance Formula (Euclidean Distance)
d = √((x2 - x1)^2 + (y2 - y1)^2)

#### Advantages
- Simple and easy to understand
- Fast for large datasets
- Efficient clustering

### Disadvantages
- Need to specify K value
- Sensitive to outliers
- Not suitable for complex shapes

### Applications
- Customer segmentation
- Image compression
- Recommendation systems

```python
# Import necessary libraries
import numpy as np              # For numerical operations
import pandas as pd             # For data handling
import matplotlib.pyplot as plt # For data visualization
```

```python
# Load the dataset
dataset = pd.read_csv('Social_Network_Ads.csv')

# Display first 5 rows
print(dataset.head())
```

```python
# Selecting only required columns (Age and Estimated Salary)
# Index 2 = Age, Index 3 = Salary
X = dataset.iloc[:, [2, 3]].values

# Display selected data
print(X[:5])
```

```python
# Import KMeans from sklearn
from sklearn.cluster import KMeans

# Create KMeans object
# n_clusters = number of groups (K)
# random_state = for same output every time
kmeans = KMeans(n_clusters=3, random_state=0)

# Fit model and predict clusters
y_kmeans = kmeans.fit_predict(X)

# Display cluster labels
print(y_kmeans)
```

```python
# Plotting data points for each cluster

# Cluster 1
plt.scatter(X[y_kmeans == 0, 0], X[y_kmeans == 0, 1],
            label='Cluster 1')

# Cluster 2
plt.scatter(X[y_kmeans == 1, 0], X[y_kmeans == 1, 1],
            label='Cluster 2')

# Cluster 3
plt.scatter(X[y_kmeans == 2, 0], X[y_kmeans == 2, 1],
            label='Cluster 3')

# Plot centroids (cluster centers)
plt.scatter(kmeans.cluster_centers_[:, 0],
            kmeans.cluster_centers_[:, 1],
            s=200, label='Centroids')

# Labels and title
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.title('K-Means Clustering')

# Show legend and graph
plt.legend()
plt.show()
```
