### K-Medoids Clustering

#### Introduction
K-Medoids is an Unsupervised Machine Learning Algorithm used for clustering.

It is similar to K-Means but more robust to noise and outliers.

#### Definition
K-Medoids divides data into K clusters by choosing actual data points as centers (medoids).

#### Key Difference (K-Means vs K-Medoids)
- K-Means → Uses mean (average)
- K-Medoids → Uses actual data point (medoid)

#### Working Steps
1. Choose value of K
2. Select K random medoids from dataset
3. Assign each data point to nearest medoid
4. Swap medoid with other points if cost reduces
5. Repeat until no change

#### Distance Measure
Can use:
- Euclidean Distance
- Manhattan Distance

#### Advantages
- Robust to outliers
- Works well with noisy data
- Uses actual data points as centers

#### Disadvantages
- Slower than K-Means
- More computational cost

#### Applications
- Customer segmentation
- Fraud detection
- Bioinformatics

```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Import KMedoids
from sklearn_extra.cluster import KMedoids
```

```python
# Load dataset
dataset = pd.read_csv('Social_Network_Ads.csv')

# Show first 5 rows
print(dataset.head())
```

```python
# Select Age and Estimated Salary
X = dataset.iloc[:, [2, 3]].values

# Display sample data
print(X[:5])
```

```python
# Create KMedoids object
# n_clusters = number of clusters
kmedoids = KMedoids(n_clusters=3, random_state=0)

# Fit model and predict clusters
y_kmedoids = kmedoids.fit_predict(X)

# Print cluster labels
print(y_kmedoids)
```

```python
# Plot clusters

# Cluster 1
plt.scatter(X[y_kmedoids == 0, 0], X[y_kmedoids == 0, 1],
            label='Cluster 1')

# Cluster 2
plt.scatter(X[y_kmedoids == 1, 0], X[y_kmedoids == 1, 1],
            label='Cluster 2')

# Cluster 3
plt.scatter(X[y_kmedoids == 2, 0], X[y_kmedoids == 2, 1],
            label='Cluster 3')

# Plot Medoids (actual data points)
plt.scatter(kmedoids.cluster_centers_[:, 0],
            kmedoids.cluster_centers_[:, 1],
            s=200, label='Medoids')

# Labels and title
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.title('K-Medoids Clustering')

# Show legend
plt.legend()
plt.show()
```

| Feature  | K-Means   | K-Medoids         |
| -------- | --------- | ----------------- |
| Center   | Mean      | Actual data point |
| Outliers | Sensitive | Robust            |
| Speed    | Fast      | Slow              |
